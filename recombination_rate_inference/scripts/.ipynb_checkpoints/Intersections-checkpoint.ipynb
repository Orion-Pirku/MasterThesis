```python
import pandas as pd
import numpy as np
import scipy as sp
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
from pybedtools import BedTool
from glob import glob
from multiprocessing import Pool, cpu_count
from typing import List, Tuple, Dict, Mapping, Union
import pyranges as pr
import numpy.typing as npt
from functools import partial
```

```python

def read_bed_file(file: str) -> pd.DataFrame:
    return pd.read_csv(file, sep = '\t', header = None, names=['Chromosome', 'Start', 'End', 'Score', 'Mindpoint'], skiprows = 1)

def parse_and_merge_bed(input_files: str) -> BedTool:
    file_paths = glob(input_files, recursive=True)
    with Pool(processes=cpu_count() // 2) as pool:
        dfs = pool.map(read_bed_file, file_paths)
    merged_df = pd.concat(dfs, ignore_index=True).sort_values(by=["Chromosome", "Start"]).reset_index().iloc[:, 1:]
    return pr.PyRanges(merged_df)

rec_rate_10kb = parse_and_merge_bed("../output/transformed_*_w10000.tsv")
print(rec_rate_10kb)
```

```python
def transform_feature_table(
    gff_object: Union[pd.DataFrame, BedTool],
    chr_pattern: str = r"chr([1-9]|[12][0-9]|3[0-3])$"
) -> BedTool:
    if isinstance(gff_object, BedTool):
        gff = gff_object.to_dataframe()
    elif isinstance(gff_object, pd.DataFrame):
        gff = gff_object.copy()
    else:
        raise TypeError("Input must be a pandas DataFrame or a BedTool object")

    required_columns: List[str] = ["chromosome", "start", "end", "# feature", "name", "symbol"]
    missing = set(required_columns) - set(gff.columns)
    if missing:
        raise ValueError(f"Missing columns: {missing}")

    df_transformed = (
        gff
        .filter(items=required_columns)
        .assign(
            start=lambda df: df["start"].astype("int64"),
            end=lambda df: df["end"].astype("int64"),
            chromosome=lambda df: "chr" + df["chromosome"].astype(str)
        )
        .loc[lambda df: df["chromosome"].str.match(chr_pattern)]
    )

    # Rename 'chromosome' to 'chrom' because BedTool expects 'chrom' as the first column
    df_transformed = df_transformed.rename(columns={"chromosome": "Chromosome", "start": "Start", "end": "End", "# feature": "Feature", "symbol": "Symbol", "name": "Name"})

    # Reorder columns so that chrom, start, end are first 3 columns (required for BedTool)
    bedtool_columns = ["Chromosome", "Start", "End", "Feature", "Symbol"]
    df_transformed = df_transformed[bedtool_columns]

    return pr.PyRanges(df_transformed.sort_values(by=['Chromosome', 'Start']).reset_index().iloc[:, 1:])

blackcap_feature_table = pd.read_csv("../input/GCF_009819655.1_bSylAtr1.pri_feature_table.txt", sep = '\t', header = 0, dtype=str)

blackcap_gff = transform_feature_table(blackcap_feature_table)
blackcap_gff.head()
```

```python
def split_bedtool_by_feature(input_bed: BedTool | pd.DataFrame) -> dict[str, BedTool]:
    
    if isinstance(input_bed, BedTool):
        df = input_bed.to_dataframe()
    elif isinstance(input_bed, pd.DataFrame):
        df = input_bed.copy()
    else:
        raise TypeError("Input must be a pandas DataFrame or a BedTool object")
    
    if df.shape[1] < 4:
        raise ValueError("Expected at least 4 columns in the BedTool Data Format (chrom, start, end, features)")

    feature_dict: dict[str, BedTool] = {}

    unique_features = df.iloc[:, 3].unique()
    
    for feature in unique_features:
        df_feature = df[df.iloc[:, 3]==feature]
        feature_dict[feature] = pr.PyRanges(df_feature)
    return feature_dict
    
features = split_bedtool_by_feature(blackcap_gff)
features
```

```python
rho_gff_intersections = (
    rec_rate_10kb.intersect(blackcap_gff, wb=True)
    .sort()
    .to_dataframe()
    .rename(columns={"name": "cM_Mb",
                     "itemRgb": "feature",
                     "blockCount": "symbol",
                     "blockSizes": "description"
                    })
        )

rho_gff_intersections.head()
rho_gff_intersections['rho_strength'] = pd.qcut(rho_gff_intersections["cM_Mb"], q=3, labels=['low','medium','high']).astype("category")
rho_gff_intersections.head()
```

```python
feature_counts = (
    rho_gff_intersections.groupby(['feature', 'rho_strength'])
    .size()
    .sort_values(ascending=False)
    .reset_index(name="count")
)
feature_counts["percent_intersections"] = ((feature_counts["count"] / feature_counts["count"].sum()).round(6))*100

feature_counts['feature'] = feature_counts["feature"].str.replace('_', ' ')
feature_counts.head(20)
```

```python
fig = px.bar(
    feature_counts,
    x = 'feature',
    y ='percent_intersections',
    color = "rho_strength",
    title ='% Intersections of Recombination Rate windows per Genomic Feature',
    labels = {'feature': 'Genomic Feature', 'percent_intersections': 'Percent of Intersections', 'rho_strength': 'Recombination Rate'}
)
fig.update_traces(textfont_size=16)
fig.update_layout(yaxis_title="% of Total Intersections", 
                  legend = dict(
                      font = dict(size=16)
                  ),
                 title = dict(
                     font=dict(size=20)
                 ))
fig.show(renderer="iframe")
fig.write_image("Intersections_rho_gff.png", width=800, height=600, scale=6)
```

```python
def chromsizes_to_dict(genome_file_name: str) -> dict[str, int]:
    genome = pd.read_csv(genome_file_name, sep="\t", header=None)
    return dict(zip(genome[0], genome[1]))

def single_iter(args):
    bed1, bed2, n_samples = args
    bed1_sub = bed1.sample(n=n_samples)
    bed2_sub = bed2.sample(n=n_samples, replace=True)
    return bed1_sub.stats.jaccard(bed2_sub)

def bootstrap_intervals(bed1: pr.PyRanges, bed2: pr.PyRanges, n_samples: int, iterations: int) -> np.ndarray:
    with Pool(processes=cpu_count() // 2) as pool:
        # Prepare arguments list: one tuple per iteration
        args = [(bed1, bed2, n_samples)] * iterations
        results = pool.map(single_iter, args)
    return np.array(results, dtype=np.float64)

def to_pyranges(obj: Union[pr.PyRanges, BedTool, pd.DataFrame]) -> pr.PyRanges:
    """Normalize input to PyRanges."""
    if isinstance(obj, pr.PyRanges):
        return obj
    elif isinstance(obj, BedTool):
        return pr.PyRanges(obj.to_dataframe())
    elif isinstance(obj, pd.DataFrame):
        return pr.PyRanges(obj)
    else:
        raise TypeError("Input must be PyRanges, BedTool, or pandas DataFrame")

def compute_jaccard(
    bed_input1: Union[pr.PyRanges, BedTool, pd.DataFrame],
    bed_input2: Union[pr.PyRanges, BedTool, pd.DataFrame, Mapping[str, Union[pr.PyRanges, BedTool, pd.DataFrame]]],
    genome_file_name: str,
    n_bootstraped_samples: int = 1000,
    iterations: int = 1000
) -> pd.DataFrame:
    
    bed1 = to_pyranges(bed_input1)
    
    # Normalize dict values if dict, else normalize directly
    if isinstance(bed_input2, Mapping):
        bed2_dict = {k: to_pyranges(v) for k, v in bed_input2.items()}
    else:
        bed2_dict = {"All": to_pyranges(bed_input2)}
    
    # Load chrom sizes (if you want to use, currently unused)
    chromsizes = chromsizes_to_dict(genome_file_name)
    
    def summarize_jaccard(jaccard_vals: np.ndarray, feature: str) -> dict:
        return {
        "feature": feature,
        "jaccard_mean": np.mean(jaccard_vals),
        "jaccard_median": np.median(jaccard_vals),
        "ci_lower_95%": np.percentile(jaccard_vals, 2.5),
        "ci_upper_95%": np.percentile(jaccard_vals, 97.5),
        }

    records = [
        summarize_jaccard(bootstrap_intervals(bed1, bed2, n_bootstraped_samples, iterations), feature) for feature, bed2 in bed2_dict.items()
    ]

    return pd.DataFrame.from_records(records)

jaccard_results = compute_jaccard(rec_rate_10kb, features,  n_bootstraped_samples= 1000, iterations=10000, genome_file_name = "../input/blackcap.genome")
```

```python

# Calculate error bar lengths relative to the mean
jaccard_results['error_lower'] = jaccard_results['jaccard_mean'] - jaccard_results['ci_lower_95%']
jaccard_results['error_upper'] = jaccard_results['ci_upper_95%'] - jaccard_results['jaccard_mean']

fig = px.bar(
    jaccard_results,
    x="feature",
    y="jaccard_mean",
    error_y=jaccard_results["error_upper"],
    error_y_minus=jaccard_results["error_lower"],
    title="Jaccard Mean with 95% Confidence Intervals by Feature",
    labels={"feature": "Feature", "jaccard_mean": "Jaccard Mean"},
    color="feature",
    color_discrete_sequence=px.colors.qualitative.Plotly,
    width=900,        # width in pixels
    height=600        # height in pixels
)

fig.update_layout(
    yaxis=dict(range=[0, 0.5]),
    xaxis_tickangle=-45,
)

fig.show()
```
